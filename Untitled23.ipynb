```python
# -*- coding: utf-8 -*-
# ===============================
# ANTI-CRASH ENVIRONMENT & MEMORY MANAGEMENT
# ===============================
import os
import gc 
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2' 
os.environ['KMP_DUPLICATE_LIB_OK'] = 'True' 

import re, numpy as np, warnings
import tensorflow as tf
import whisper
import matplotlib.pyplot as plt
from moviepy import VideoFileClip
from tensorflow.keras import layers, models, regularizers
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

warnings.filterwarnings("ignore")

# ===============================
# CONFIG M1 (TEXT ONLY)
# ===============================
# Path utama (Ganti sesuai lokasi folder Dataset Anda)
BASE_DATASET = r"/var/www/html/video_clasification_fusion/Dataset" 

PROCESSED_DATA_FILE = "processed_data_text_676.npz" 
MODEL_FILE = "best_model_m1_text_676.h5"

MAX_TEXT_LEN = 100
BATCH_SIZE = 8
EPOCHS = 30 
MAX_SAMPLES_PER_CLASS = 676 

# ===============================
# LOAD MODELS
# ===============================
print("Initializing AI Engines (Whisper STT)...")
stt_model = whisper.load_model("base")

# ===============================
# HELPER FUNCTIONS
# ===============================
def find_category_dir(base_path, target_name):
    """Mencari nama folder tanpa memperdulikan huruf besar/kecil (Case-insensitive)"""
    if not os.path.exists(base_path): return None
    for item in os.listdir(base_path):
        if item.lower() == target_name.lower() and os.path.isdir(os.path.join(base_path, item)):
            return os.path.join(base_path, item)
    return None

def extract_text_from_video(video_path):
    """Mengekstrak audio dari video dan mengubahnya menjadi teks menggunakan Whisper"""
    tmp = f"temp_proc_{os.getpid()}.wav"
    try:
        with VideoFileClip(video_path) as video:
            if video.audio is None: 
                video.close()
                return ""
            video.audio.write_audiofile(tmp, logger=None)
            video.close() 
            
        text = ""
        try: 
            text = stt_model.transcribe(tmp)['text']
        except: 
            pass
        
        if os.path.exists(tmp): os.remove(tmp)
        return text.lower()
    except Exception:
        if os.path.exists(tmp): os.remove(tmp)
        return ""

def text_to_sequence(text):
    """Konversi teks ke sekuens angka (ASCII mod 5000)"""
    text = re.sub(r'[^a-zA-Z ]', '', text)
    seq = [ord(c) % 5000 for c in text][:MAX_TEXT_LEN]
    seq += [0] * (MAX_TEXT_LEN - len(seq))
    return np.array(seq)

# ===============================
# DATA LOADER (TEKS SAJA & CACHE VALIDATED)
# ===============================
def load_dataset_text():
    # 1. Validasi Cache: Hapus otomatis jika cache berisi 0 sampel
    if os.path.exists(PROCESSED_DATA_FILE):
        d = np.load(PROCESSED_DATA_FILE)
        X_loaded, y_loaded = d['X_txt'], d['y']
        if len(X_loaded) > 0:
            print(f"[INFO] Memuat {len(X_loaded)} data dari cache: {PROCESSED_DATA_FILE}")
            return X_loaded, y_loaded
        else:
            print(f"[WARNING] Cache '{PROCESSED_DATA_FILE}' terdeteksi kosong (0 data). Menghapus cache dan membaca ulang dataset...")
            os.remove(PROCESSED_DATA_FILE)
    
    # 2. Check & Fallback Path Dataset
    path_to_use = BASE_DATASET
    if not os.path.exists(path_to_use):
        local_dataset = os.path.join(os.getcwd(), "Dataset")
        if os.path.exists(local_dataset):
            print(f"[INFO] BASE_DATASET disesuaikan ke folder lokal: '{local_dataset}'")
            path_to_use = local_dataset
        else:
            raise FileNotFoundError(
                f"\n[ERROR] Folder dataset tidak ditemukan di '{BASE_DATASET}' maupun '{local_dataset}'!"
                f"\nPastikan path folder Dataset sudah benar."
            )

    print(f"Mengekstrak data TEKS dari: {path_to_use}")
    total_target = MAX_SAMPLES_PER_CLASS * 2
    
    X_t = np.zeros((total_target, MAX_TEXT_LEN), dtype=np.int32)
    Y = np.zeros((total_target,), dtype=np.int32)
    
    categories = {'nonkasar': 0, 'kasar': 1}
    idx = 0 
    
    for cat_name, label in categories.items():
        cat_path = find_category_dir(path_to_use, cat_name)
        if not cat_path:
            print(f"[WARNING] Folder '{cat_name}' tidak ditemukan di {path_to_use}")
            continue
        
        subfolders = [d for d in os.listdir(cat_path) if os.path.isdir(os.path.join(cat_path, d)) and not d.startswith('.')]
        if not subfolders: subfolders = [""]
            
        quota_per_subfolder = MAX_SAMPLES_PER_CLASS // len(subfolders)
        remainder = MAX_SAMPLES_PER_CLASS % len(subfolders)
        class_count = 0 
        
        for sf_idx, sf in enumerate(subfolders):
            sf_path = os.path.join(cat_path, sf) if sf else cat_path
            current_quota = quota_per_subfolder + (1 if sf_idx < remainder else 0)
            sub_count = 0
            
            # Rekursif menggunakan os.walk untuk memastikan semua video ditemukan
            for root, dirs, files in os.walk(sf_path):
                if ".ipynb_checkpoints" in root: continue
                for f in files:
                    if class_count >= MAX_SAMPLES_PER_CLASS: break 
                    if sub_count >= current_quota: break 
                    
                    if f.lower().endswith((".mp4", ".avi", ".mkv", ".mov")):
                        full_path = os.path.join(root, f)
                        try:
                            txt_a = extract_text_from_video(full_path)
                            
                            X_t[idx] = text_to_sequence(txt_a)
                            Y[idx] = label
                            
                            sub_count += 1
                            class_count += 1
                            idx += 1
                            
                            folder_display = sf if sf else cat_name
                            print(f"[TEKS | {cat_name.upper()} | {folder_display}] {sub_count}/{current_quota} (Total: {class_count}/{MAX_SAMPLES_PER_CLASS}) LOADED: {f[:25]}")
                            
                            del txt_a
                            if class_count % 20 == 0: gc.collect()
                        except Exception as e:
                            print(f"[ERROR READ] Gagal membaca {f}: {e}")
                            continue
                if class_count >= MAX_SAMPLES_PER_CLASS or sub_count >= current_quota:
                    break

    X_t, Y = X_t[:idx], Y[:idx]

    if len(X_t) == 0:
        raise ValueError(
            f"\n[ERROR] 0 file video dimuat dari '{path_to_use}'!"
            "\nPastikan folder Kasar dan Nonkasar berisi file video berformat .mp4/.avi."
        )

    np.savez(PROCESSED_DATA_FILE, X_txt=X_t, y=Y)
    return X_t, Y

# ===============================
# MODEL M1 (UNIMODAL TEKS)
# ===============================
def build_text_model():
    in_txt = layers.Input(shape=(MAX_TEXT_LEN,))
    t = layers.Embedding(5000, 64)(in_txt)
    t = layers.SpatialDropout1D(0.2)(t)
    t = layers.Bidirectional(layers.GRU(64))(t)
    t = layers.Dense(128, activation='relu')(t)

    x = layers.Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.02))(t)
    x = layers.Dropout(0.5)(x)
    x = layers.BatchNormalization()(x)
    out = layers.Dense(1, activation='sigmoid', name='prediction')(x)

    model = models.Model(in_txt, out)
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=2e-4), 
                  loss=tf.keras.losses.BinaryCrossentropy(label_smoothing=0.1), 
                  metrics=['accuracy'])
    return model

# ===============================
# MAIN & VISUALIZATION
# ===============================
def main():
    X_t, Y = load_dataset_text()
    X_t_tr, X_t_val, Y_tr, Y_val = train_test_split(
        X_t, Y, test_size=0.2, stratify=Y, random_state=42)
    
    if not os.path.exists(MODEL_FILE):
        print("\nTraining Model M1 (Baseline Unimodal Teks)...")
        model = build_text_model()
        
        callbacks = [
            tf.keras.callbacks.EarlyStopping(patience=8, restore_best_weights=True),
            tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-6)
        ]
        
        history = model.fit(
            X_t_tr, Y_tr,
            validation_data=(X_t_val, Y_val),
            epochs=EPOCHS, batch_size=BATCH_SIZE,
            callbacks=callbacks
        )
        model.save(MODEL_FILE)
        
        plt.figure(figsize=(12, 4))
        plt.subplot(1, 2, 1); plt.plot(history.history['accuracy'], label='Train'); plt.plot(history.history['val_accuracy'], label='Val'); plt.title('M1 Text Acc'); plt.legend()
        plt.subplot(1, 2, 2); plt.plot(history.history['loss'], label='Train'); plt.plot(history.history['val_loss'], label='Val'); plt.title('M1 Text Loss'); plt.legend()
        plt.show()
    else:
        model = tf.keras.models.load_model(MODEL_FILE)

    # Evaluasi Akhir
    preds = model.predict(X_t_val)
    print("\n--- Final Report (M1 - Teks Saja) ---")
    print(classification_report(Y_val, (preds > 0.5).astype(int)))
    
    # Confusion Matrix
    cm = confusion_matrix(Y_val, (preds > 0.5).astype(int))
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Safe', 'Harsh'], yticklabels=['Safe', 'Harsh'])
    plt.title('Confusion Matrix M1 (Teks)')
    plt.show()

    # Testing 10 File per Sub-folder
    path_to_use = BASE_DATASET if os.path.exists(BASE_DATASET) else os.path.join(os.getcwd(), "Dataset")
    print("\n>>> TESTING 10 FILES PER SUB-FOLDER (TEKS SAJA)")
    for main_cat in ['kasar', 'nonkasar']:
        cat_path = find_category_dir(path_to_use, main_cat)
        if not cat_path: continue
        
        subfolders = [d for d in os.listdir(cat_path) if os.path.isdir(os.path.join(cat_path, d)) and not d.startswith('.')]
        if not subfolders: subfolders = [""]
        
        for sub in subfolders:
            sub_full = os.path.join(cat_path, sub) if sub else cat_path
            files = [f for f in os.listdir(sub_full) if f.lower().endswith((".mp4", ".avi", ".mkv", ".mov"))][:10]
            if not files: continue
            
            folder_title = sub if sub else main_cat
            print("="*80)
            print(f"Folder:  {folder_title}")
            print("="*80)
            
            for i, fname in enumerate(files, 1):
                txt_a = extract_text_from_video(os.path.join(sub_full, fname))
                
                pred = model.predict(np.expand_dims(text_to_sequence(txt_a), 0), verbose=0)
                
                p = pred[0][0]
                status = 'Kasar' if p >= 0.5 else 'Aman '
                
                print(f"{i:2}. [{status}]  {p*100:6.2f}%   | {fname[:45]}")
                if txt_a.strip():
                    print(f"    ↳ Teks Deteksi : '{txt_a}'")
                
                del txt_a
                gc.collect()

if __name__ == "__main__":
    main()
```
